```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("../../datasets/advertising.csv")
```

```python
from linear_regression import LinearRegression
```

```python
X=df["TV"].values
y=df["Sales"].values
```

```python
X_mean=X.mean()
X_std=X.std()
X=(X-X_mean)/X_std
#Applied Normalization for Gradient Descent Function
```

```python
model= LinearRegression(
       learning_rate=0.0001,
       epochs=50000
)
```

```python
model.fit(X,y)
```

```python
y_pred=model.predict(X)
```

```python
print(model.w)
print(model.b)
```

```python
plt.scatter(X,y)
plt.plot(X,y_pred,color="red")
plt.show()
```

```python
from sklearn.metrics import r2_score

y_pred = model.predict(X)

r2 = r2_score(y, y_pred)
print(f"R² Score: {r2:.4f}")
```
